```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
metadata = pd.read_csv("/dodo/rdd4/bygul_simulations_bronko/rsv_nextstrain/rsva_nextstrain_metadata_08192026.tsv", sep="\t")
metadata
```

```python
metadata.columns
```

```python
## Quality control for full coverage sequences
complete_genomes_metadata = metadata[(metadata['genome_coverage'] == 1.0) & (metadata['missing_data'] == 0)]
complete_genomes_metadata = complete_genomes_metadata[~complete_genomes_metadata['date'].str.contains('X')]
complete_genomes_metadata
```

```python
## partition by month and clade
complete_genomes_metadata['month'] = pd.to_datetime(complete_genomes_metadata['date']).dt.to_period('M')
```

```python
##group by month and clade and return first
selected = complete_genomes_metadata.sort_values(by='date')[['month', 'clade', 'INSDC_accession', 'PPX_accession']].groupby(['month', 'clade']).first().reset_index()
selected
```

```python
from Bio import SeqIO

fasta_file = "/dodo/rdd4/bygul_simulations_bronko/rsv_nextstrain/rsva_nextstrain_seqs_08192026.fasta"
output_dir = "/dodo/rdd4/bygul_simulations_bronko/rsv_nextstrain/"

# Load FASTA sequences into a dictionary
records = {
    record.id.split(".")[0]: record
    for record in SeqIO.parse(fasta_file, "fasta")
}

for _, row in selected.iterrows():
    accession = row["PPX_accession"].split(".")[0]

    if accession not in records:
        print(f"Warning: {accession} not found")
        continue

    # Header / filename
    label = f"{accession}_{row['clade']}_{row['month']}"

    # Make a copy so we don't modify the original record
    record = records[accession][:]
    record.id = label
    record.name = label
    record.description = ""

    # Write individual FASTA
    SeqIO.write(
        record,
        f"{output_dir}{label}.fasta",
        "fasta"
    )
```
